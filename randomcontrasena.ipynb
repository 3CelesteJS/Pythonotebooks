```python
import string
import random
```

```python
longitud=(input("Ingrese el tamaño de la contraseña:"))
```

```python
caracteres= (string.ascii_letters + string.digits + string.punctuation)
```

```python
print(caracteres)
```

```python
contrasena= "".join(random.choice(caracteres)for i in range(int(longitud)))
```

```python
print("La contraseña generada es: " + contrasena)
```
